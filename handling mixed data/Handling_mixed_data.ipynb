```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('titanic modified.csv')
```

```python
df
```

```python
df['number'].unique()
```

```python
fig=df['number'].value_counts().plot(kind='bar')
fig.set_title('passanger traveling with number')
fig.set_xlabel('number')
fig.set_ylabel('count')
```

```python
#extract numerical part
df['number_numerical']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

np.where(...) → For each row:

If number_numerical is NaN, use df['number'] (likely a categorical/original version)

Else, set to NaN

```python
#extract  categorical part
df['number_categorical']=np.where(df['number_numerical'].isnull(),df['number'],np.nan)
df.head()
```

```python
#now handling cabin and ticket column
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
df['cabin_num']=df['Cabin'].str.extract('(\d+)') #capture numeric part
df['cabin_cat']=df['Cabin'].str[0] #capture first letter
```

```python
df.head()
```

```python
df["cabin_cat"].value_counts()
```

```python
#extracting last bit of ticket as number
df['ticket_num']=df['Ticket'].apply(lambda s:s.split()[-1])
df['ticket_num']=pd.to_numeric(df['ticket_num'],errors='coerce',downcast='integer')

#extracting the first part of ticket as category
df['ticket_cat']=df['Ticket'].apply(lambda s:s.split()[0])
df['ticket_cat']=np.where(df['ticket_cat'].str.isdigit(),np.nan,df['ticket_cat'])

df.head()
```
