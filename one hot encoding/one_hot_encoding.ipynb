one hot encoding using sklearn

```python
df.head()
```

```python
df2=df.iloc[:,0:4]
df2
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
 #we use sparse=false as now it will return numpy array
#to drop 1 column from both fuel and owner use  ohe=OneHotEncoder(drop='first',sparse='False')
# if you dont add dtype bydefault all values will be float
ohe = OneHotEncoder(sparse_output=False,dtype=np.int32)
```

```python
df2_new=ohe.fit_transform(df2[['fuel','owner']])
```

```python
#this new df2_new only have dummy variable of fuel and owner brand and km_driven we have to append
df2_new.shape
```

```python
np.hstack([df2[['brand','km_driven']],df2_new])
```

One hot encoding for multiple category

```python
counts=df['brand'].value_counts()
```

```python
threshold=100
```

```python
rep1=counts[counts<=threshold].index
```

```python
#cars less than 100 are categories as uncommon
pd.get_dummies(df['brand'].replace(rep1,'uncommon'))
```
